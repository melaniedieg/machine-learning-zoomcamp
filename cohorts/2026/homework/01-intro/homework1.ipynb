```python
asia_7 = asia[['vehicle_weight', 'model_year']].head(7)
```

```python
#convert to array

X = asia_7.values
```

```python
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
w
```
